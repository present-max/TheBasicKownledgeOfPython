```python
import pandas as pd


df=pd.read_csv("resourse/movie_top100.csv")
df.head(10)#显示前10行数据
df.tail(10)#显示后10行数据
df.describe()#显示数据的统计信息
df["电影名"]#选择列
df[["电影名","导演","上映时间","评分"]]#选择多列
df.iloc[0:5]#基于行号选择行
df.loc[0:5]#基于行索引号选择行和列

```

```python
#条件过滤
df[df["评分"]>8.5]
df[df["类型"].isin(["动画","喜剧"])]
df[df["评分"].between(80,90)]#包含80和90，也可以自己设置那边包含
df[(df["评分"]>8.5) & (df["类型"].isin(["动画","喜剧"])) & (df["评分"].between(80,90))]#多条件过滤
df[(df["评分"]>8.5) | (df["类型"].isin(["动画","喜剧"])) | (df["评分"].between(80,90))]


```

```python
#数据清洗
df.dropna()#删除有空值的行，返回新的DataFrame，原DataFrame不变
df.dropna(axis=1)#删除有空值的列
df.drop(0)#删除指定索引行
df.fillna(0)#用0填充空值
df.replace(0,1)#用1替换0
df.duplicated()#检测重复行
df.drop_duplicates()#删除重复行（整行重复才删除）
df.drop_duplicates(subset=["电影名"],keep="first")#删除重复行，subset指定根据哪一列进行去重，keep指定保留哪个，first表示保留第一个，last表示保留最后一个，False表示不保留


```

```python
#数据排序
df.sort_values(by="评分",ascending=False)#按评分降序排序
df.sort_values(by="评分",ascending=True)#按评分升序排序
df.sort_values(by=["评分","上映时间"],ascending=[False,True])#按评分降序排序，如果评分相同则按上映时间升序排序

```

```python
#数据分组
df.groupby("语言")["评分"].mean()
df.groupby("语言")["评分"].max()
df.groupby("语言")["评分"].min()
df.groupby("语言")["评分"].sum()
df.groupby("语言")["评分"].count()



```
